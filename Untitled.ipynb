```python
# ==========================================
# TUGAS METODE NUMERIK
# Penyelesaian SPL dengan Metode Gauss-Jordan
# Nama : Nadia Hasanah
# NIM  : 20250040113
# ==========================================

from fractions import Fraction

# Fungsi untuk menyelesaikan sistem persamaan linear
# menggunakan metode eliminasi Gauss-Jordan.
def gauss_jordan(matrix):

    # Menghitung jumlah persamaan dalam matriks
    n = len(matrix)

    # Mengubah nilai matriks menjadi bentuk pecahan
    # agar hasil perhitungan lebih akurat.
    A = [[Fraction(x) for x in row] for row in matrix]

    # Melakukan proses eliminasi pada setiap kolom
    for i in range(n):

        # Mengecek apakah elemen diagonal bernilai nol.
        # Jika nol, maka baris akan ditukar.
        if A[i][i] == 0:
            for j in range(i + 1, n):
                if A[j][i] != 0:
                    A[i], A[j] = A[j], A[i]
                    break

        # Menentukan nilai pivot pada baris yang sedang diproses
        pivot = A[i][i]

        # Membagi seluruh elemen pada baris dengan nilai pivot
        # supaya elemen diagonal menjadi 1.
        for j in range(n + 1):
            A[i][j] = A[i][j] / pivot

        # Mengubah elemen lain pada kolom yang sama menjadi nol
        for k in range(n):
            if k != i:
                faktor = A[k][i]

                # Mengurangi baris lain menggunakan baris pivot
                for j in range(n + 1):
                    A[k][j] -= faktor * A[i][j]

    # Mengambil hasil akhir dari masing-masing variabel
    hasil = [A[i][n] for i in range(n)]

    return hasil


# ==========================================
# A. PEMODELAN DAN SOLUSI AWAL
# ==========================================

print("=== A. PEMODELAN DAN SOLUSI AWAL ===")

# Memasukkan persamaan berdasarkan tabel pada soal.
# Baris pertama merupakan persamaan CPU.
# Baris kedua merupakan persamaan RAM.
# Baris ketiga merupakan persamaan Bandwidth.
matriks_awal = [
    [2, 1, 1, 13],
    [4, 3, 3, 31],
    [8, 7, 9, 71]
]

# Memanggil fungsi Gauss-Jordan untuk mencari nilai variabel
hasil_awal = gauss_jordan(matriks_awal)

# Menyimpan hasil ke dalam variabel masing-masing
x1, x2, x3 = hasil_awal

# Menampilkan hasil perhitungan awal
print("Jumlah Web      (x1) =", x1)
print("Jumlah API      (x2) =", x2)
print("Jumlah Database (x3) =", x3)


# ==========================================
# VERIFIKASI HASIL AWAL
# ==========================================

print("\n=== VERIFIKASI HASIL AWAL ===")

# Memasukkan kembali nilai hasil ke persamaan awal
# untuk memastikan semua resource digunakan sesuai kapasitas.
cpu = 2*x1 + x2 + x3
ram = 4*x1 + 3*x2 + 3*x3
bandwidth = 8*x1 + 7*x2 + 9*x3

# Menampilkan hasil verifikasi
print("CPU       =", cpu)
print("RAM       =", ram)
print("Bandwidth =", bandwidth)


# ==========================================
# B. ANALISIS SENSITIVITAS
# ==========================================

print("\n=== B. ANALISIS SENSITIVITAS ===")

# Mengambil digit terakhir NIM, yaitu 3
d = 3

# Menghitung bandwidth baru berdasarkan rumus 71 + d
bandwidth_baru = 71 + d

print("Digit terakhir NIM =", d)
print("Bandwidth baru     =", bandwidth_baru, "Mbps")

# Membuat matriks baru dengan bandwidth yang sudah berubah.
# Persamaan CPU dan RAM tetap, hanya bandwidth yang berubah.
matriks_baru = [
    [2, 1, 1, 13],
    [4, 3, 3, 31],
    [8, 7, 9, bandwidth_baru]
]

# Menghitung kembali SPL menggunakan metode Gauss-Jordan
hasil_baru = gauss_jordan(matriks_baru)

# Menyimpan hasil perhitungan sensitivitas
x1_baru, x2_baru, x3_baru = hasil_baru

# Menampilkan hasil perhitungan yang baru
print("\nHasil setelah perubahan bandwidth:")
print("Jumlah Web      (x1) =", x1_baru)
print("Jumlah API      (x2) =", x2_baru)
print("Jumlah Database (x3) =", x3_baru)


# ==========================================
# VERIFIKASI HASIL BARU
# ==========================================

print("\n=== VERIFIKASI HASIL BARU ===")

# Menghitung kembali penggunaan resource
# berdasarkan hasil perhitungan sensitivitas.
cpu_baru = 2*x1_baru + x2_baru + x3_baru
ram_baru = 4*x1_baru + 3*x2_baru + 3*x3_baru
bw_baru = 8*x1_baru + 7*x2_baru + 9*x3_baru

# Menampilkan hasil verifikasi resource baru
print("CPU       =", cpu_baru)
print("RAM       =", ram_baru)
print("Bandwidth =", bw_baru)


# ==========================================
# KESIMPULAN KELAYAKAN OPERASIONAL
# ==========================================

print("\n=== KESIMPULAN ===")

# Mengecek apakah seluruh hasil berupa bilangan bulat
# dan tidak memiliki nilai negatif.
if all(x.denominator == 1 and x >= 0 for x in hasil_baru):

    print("Solusi baru layak secara operasional.")
    print("Semua instance berupa bilangan bulat tak negatif.")

else:

    print("Solusi baru tidak layak secara operasional.")
    print("Terdapat jumlah instance yang berbentuk pecahan.")
```
